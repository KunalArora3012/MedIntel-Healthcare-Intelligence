```python
dfs[2][1]['START']=pd.to_datetime(dfs[2][1]['START'])
dfs[2][1]['STOP']=pd.to_datetime(dfs[2][1]['STOP'])
```

```python
Time_wise_patients = dfs[2][1].groupby(
    [dfs[2][1]['START'].dt.year,dfs[2][1]['START'].dt.month]).agg(
    {'PATIENT':'nunique'})
```

```python
This_month = Time_wise_patients.iloc[-2,-1]  
Last_month = Time_wise_patients.iloc[-3,-1]
Growth =( (This_month / Last_month )-1 )*100
print(str(round(Growth))+'%')
```

```python
# Total Encounters
dfs[2][1]['Id'].nunique()
```

```python
Time_wise_Encounters = dfs[2][1].groupby(
    [dfs[2][1]['START'].dt.year,dfs[2][1]['START'].dt.month]).agg(
    {'Id':'nunique'})
This_monthE = Time_wise_Encounters.iloc[-2,-1]  
Last_monthE = Time_wise_Encounters.iloc[-3,-1]
GrowthE =( (This_monthE / Last_monthE )-1 )*100
print(str(round(GrowthE))+'%')
```

```python
years= [i[0] for i in Time_wise_Encounters.index] # extraction of Years from index
month= [i[1] for i in Time_wise_Encounters.index] # extraction of month from index
```

```python
Time_wise_Encounters['year']=years # creating new column for year
Time_wise_Encounters['month']=month # creating new column for month 
```

```python
Time_wise_Encounters=Time_wise_Encounters.reset_index(drop=True) # dropping index of time wise encounters
```

```python
ME = Time_wise_Encounters.groupby('month').agg({'Id':'sum'}).reset_index() # grouping month wise data 
px.area(ME,x='month',y='Id',markers='o',color='month') # creating line chart 
```

```python
YE = Time_wise_Encounters.groupby('year').agg({'Id':'sum'}).reset_index() # grouping year wise data 
px.area(YE,x='year',y='Id',markers='o',color='year') # creating line chart
```

```python
from datetime import datetime as dt # getting function for current date

dfs[1][1]['Age']=dt.now().year - pd.to_datetime(
    dfs[1][1]['BIRTHDATE']).dt.year # calculating age as per curret date 
```

```python
dfs[1][1]['Age_bin']=pd.cut(
    dfs[1][1]['Age'],
    bins = 5 ,
    labels=['35 to 49', '49 to 63', '63 to 76','76 to 90','90 to 104'] ) # creating bins for age 
```

```python
Age_wise_patients=dfs[1][1].groupby('Age_bin',observed=True).agg({'Id':'nunique'}).reset_index() 
# total patients as per bin
```

```python
px.bar(Age_wise_patients , x='Age_bin' , y='Id' , color='Id' )
# creating bar chart
```

```python
GD = dfs[1][1]['GENDER'].value_counts().reset_index()
```

```python
fig=px.pie( GD,
       names=GD['GENDER'],
       values=GD['count'],
       hole=0.5)
fig.add_annotation(
    text=f'Total Patients : {dfs[1][1]['Id'].nunique()} ',
    x=0.5,
    y=0.5,
    font_size=10,
    showarrow=False)

fig.show()
```

```python
Payers=pd.read_csv('payers.csv')
```

```python
Payers.head(1)
```

```python
payer=pd.merge(dfs[2][1],Payers , left_on ='PAYER',right_on = 'Id' , how ='inner')
```

```python
Top_Payers=payer.groupby('NAME').agg(
    {'PAYER_COVERAGE':'sum'}).sort_values('PAYER_COVERAGE').iloc[::,:].reset_index()
```

```python
fig=px.funnel(Top_Payers,x='PAYER_COVERAGE',y='NAME',color='NAME')
fig.update_traces(
    textposition = 'inside',
    textfont_size=14
)
```

```python
# Top Encounter Class 
dfs[2][1]['DESCRIPTION']=dfs[2][1]['DESCRIPTION'].str.replace('(procedure)','')
Top_Encounter=dfs[2][1].groupby('DESCRIPTION').agg(
    {'Id':'count'}).sort_values('Id').iloc[::,:].reset_index().tail(5)
```

```python
fig=px.funnel(Top_Encounter,x='Id',y='DESCRIPTION',color='DESCRIPTION')
fig.update_traces(
    textposition = 'inside',
   textfont_size=14
)
```

```python
Desc = dfs[2][1].groupby('ENCOUNTERCLASS').agg({'Id':'count'}).reset_index().sort_values('Id').tail(5)
```

```python
fig=px.pie( Desc,
       names='ENCOUNTERCLASS',
       values='Id',
       hole=0.5)
fig.add_annotation(
    text=f'Total Patients : {dfs[1][1]['Id'].nunique()} ',
    x=0.5,
    y=0.5,
    font_size=10,
    showarrow=False)

fig.show()
```
